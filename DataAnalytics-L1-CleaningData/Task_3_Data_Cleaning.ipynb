```python
categorical_columns = [
    'order_status',
    'country_code',
    'acquisition_channel',
    'customer_segment'
]

for col in categorical_columns:
    print(f"\n{col.upper()}")
    print(df[col].value_counts(dropna=False))
```

```python
print(df['ordered_at'].head(10))
print("\nDate data type:", df['ordered_at'].dtype)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    print(f"\n{col}")
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
```

```python
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} potential outliers")
```

```python
print(df.dtypes)
```

```python
before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

print("Rows before:", before_duplicates)
print("Rows after:", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)
```

```python
print(df.isnull().sum())
```

```python
df['order_status'] = df['order_status'].astype(str).str.strip().str.lower()

df['order_status'] = df['order_status'].replace({
    'completed': 'Completed',
    'cancelled': 'Cancelled',
    'partially_refunded': 'Partially_Refunded',
    'refunded': 'Refunded'
})

print(df['order_status'].value_counts())
```

```python
df['country_code'] = df['country_code'].str.strip().str.upper()

df['acquisition_channel'] = df['acquisition_channel'].str.strip().str.lower()

df['customer_segment'] = df['customer_segment'].str.strip().str.lower()

print("Categories standardised successfully!")
```

```python
df['ordered_at'] = pd.to_datetime(df['ordered_at'], errors='coerce')

print("Date conversion completed!")
print("New date data type:", df['ordered_at'].dtype)
print("Invalid dates:", df['ordered_at'].isna().sum())
```

```python
numeric_cols = [
    'line_count', 'units', 'subtotal',
    'tax_amount', 'shipping_amount',
    'gross_payment', 'refund_amount', 'net_revenue'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

print("Numeric columns corrected!")
print(df[numeric_cols].dtypes)
```
