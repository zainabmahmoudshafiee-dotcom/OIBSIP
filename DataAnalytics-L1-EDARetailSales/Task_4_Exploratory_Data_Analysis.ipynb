```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
import pandas as pd

# Load the correct CSV file instead of the non-existent Excel file
df = pd.read_csv('/content/ecommerce_sales_cleaned (1).csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
print("Total Net Revenue:", df['net_revenue'].sum())
print("Average Net Revenue:", df['net_revenue'].mean())
print("Highest Net Revenue:", df['net_revenue'].max())
print("Lowest Net Revenue:", df['net_revenue'].min())
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x='customer_segment', y='net_revenue')

plt.title('Net Revenue by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Net Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='acquisition_channel')

plt.title('Number of Orders by Acquisition Channel')
plt.xlabel('Acquisition Channel')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='country_code')

plt.title('Number of Orders by Country')
plt.xlabel('Country')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.show()
```

```python
df['ordered_at'] = pd.to_datetime(df['ordered_at'])

print("Date column converted successfully!")
```

```python
df['month'] = df['ordered_at'].dt.to_period('M')

monthly_revenue = df.groupby('month')['net_revenue'].sum()

monthly_revenue.head(10)
```

```python
plt.figure(figsize=(12, 5))

monthly_revenue.plot(kind='line', marker='o')

plt.title('Monthly Net Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Net Revenue')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# Calculate total refund amount by order status to fix the NameError
refund_by_status = df.groupby('order_status')['refund_amount'].sum()

plt.figure(figsize=(9, 5))
refund_by_status.plot(kind='bar', color='salmon')

plt.title('Total Refund Amount by Order Status')
plt.xlabel('Order Status')
plt.ylabel('Total Refund Amount')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(data=df, x='units', y='net_revenue')

plt.title('Relationship Between Units Sold and Net Revenue')
plt.xlabel('Units Sold')
plt.ylabel('Net Revenue')
plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of E-commerce Sales')
plt.show()
```

```python
country_revenue = df.groupby('country_code')['net_revenue'].sum().sort_values(ascending=False)

country_revenue.head(10)
```

```python
plt.figure(figsize=(10, 6))

country_revenue.head(10).plot(kind='bar')

plt.title('Top 10 Countries by Net Revenue')
plt.xlabel('Country')
plt.ylabel('Total Net Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
segment_revenue = df.groupby('customer_segment')['net_revenue'].sum().sort_values(ascending=False)

segment_revenue
```

```python
plt.figure(figsize=(10, 5))

segment_revenue.plot(kind='bar')

plt.title('Total Net Revenue by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Net Revenue')
plt.xticks(rotation=45)
plt.show()
```
